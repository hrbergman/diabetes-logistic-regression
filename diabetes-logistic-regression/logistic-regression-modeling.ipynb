## B1. 4 Assumptions of Logistic Regression Model 


#### Assumption 1: The dependent variable is binary (or ordinal).

Logistic regression is a statistical model used to predict the probability of an event occurring, based on a given set of independent variables (IBM, 2024). It's particularly useful when the dependent variable is categorical in nature, meaning it can only take on a limited number of values. 

There are two primary types of logistic regression:
- Binary logistic regression: With this type, the dependent variable has only two possible outcomes (such as "Yes" or "No", "True" or "False"). 
- Ordinal logistic regression: This second type is used when the dependent variable has ordered categories (such as "Low", "Medium", and "High"). 

Given that our dependent variable, Diabetes, is binary (with only two possible outcomes: Yes or No), we'll employ binary logistic regression for our analysis.
 

#### Assumption 2: Our observations are independent of each other. 

Logistic regression assumes that the observations in our dataset are independent of each other. This means the data points should not represent repeated measurements of the same individual or be related or matched in any way.

#### Assumption 3: There is little to no multicollinearity. 

Logistic regression assumes that there is little or no multicollinearity among our independent (explanatory) variables. This means is that our independent variables shouldn't be too highly correlated with each other. 

Multicollinearity is a condition that "occurs when two or more independent variables have a high correlation with one another in a regression model", can significantly impair the performance of our multiple linear regression model. Presence of multicollinearity makes it "difficult to determine the individual effect of each independent variable on the dependent variable", leading to inflated standard errors and unreliable coefficient estimates. Identifying multicollinearity is essential to prevent distorted results and ensure model interpretability. One effective method "to detect multicollinearity is to calculate the variance inflation factor (VIF) for each independent variable" (Bhandari, 2024). 

#### Assumption 4: There are no extreme outliers. 

Logistic regression assumes that there are no extreme outliers or influential observations in the dataset. This is because such data points can disproportionately impact the outcomes of the model, leading to biased and unreliable results.

Outliers are data points that are significantly different from the rest of the observations in the dataset. They can be caused by a variety of factors, such as measurement errors, data entry mistakes, or unusual events. Influential observations are data points that have a strong impact on the results of the model. This can occur even if the data points are not outliers.

The presence of outliers or influential observations can make it difficult to interpret the results of a logistic regression model. For example, outliers can make it appear that there is a relationship between two variables when there is none. Influential observations can make it appear that a variable is more important than it actually is (Bobbitt, 2020).

There are several options for dealing with outliers and influential observations. One option is to remove them from the dataset. This is the simplest approach, but it can also lead to a loss of information. Another option is to substitute the outliers or influential observations with a value such as the median or mean. This can help to reduce their impact on the model, but it can also introduce bias.

A third option is to retain the outliers or influential observations within the model, but to acknowledge their presence during the reporting of our regression analysis results. This allows us to be transparent about the potential impact of these data points on the model's results.

## D2. Feature Selection Procedure

### Step 1. Multicollinearity Removal (VIF Analysis) 
To mitigate any adverse effects of multicollinearity, we'll use the Variance Inflation Factor (VIF) to identify and eliminate features that exhibit high correlations with other predictors. VIF measures the extent to which the variance of a regression coefficient is inflated due to multicollinearity. As noted by Bhandari (2024), removing variables with high VIF values can effectively reduce multicollinearity, leading to improved model accuracy and stability. For this procedure, we'll consider a VIF threshold of 5 as indicative of problematic multicollinearity and remove any features that exceed this value.

```python
# Analyze Variance Inflation Factors (VIF) to determine if any variables should be removed due to strong correlations

# Function to calculate VIF for each variable
def calculate_vif(X):
    vif = pd.DataFrame() # create dataframe to store VIF results
    vif["feature"] = X.columns # features in X
    vif["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])] # use the vif function to calc VIF for ea column in X
    return vif

# Define the features
features = ['length_of_stay', 'age', 'income', 'daily_charge', 'children_in_household', 
            'overweight', 'back_pain', 'high_bp', 'stroke', 
            'gender_male', 'gender_nonbinary',
            'complication_risk_level_low', 'complication_risk_level_medium', 
            'admission_type_emergency', 'admission_type_observation']

# Subset the dataset
X1 = df_subset[features]

# Set a threshold for VIF
vif_threshold = 5

# Remove variables with high VIF
while True:
    vif_scores = calculate_vif(X1)
    
    # Get the max VIF value
    max_vif = vif_scores['VIF'].max()  
    
    # IF the max VIF is above the threshold...
    if max_vif > vif_threshold:  
        # Get the feature with the highest VIF
        feature_to_remove = vif_scores.sort_values("VIF", ascending=False).iloc[0]["feature"]
        # Display what features we're removing with their VIF 
        print(f"REMOVING VARIABLE: {feature_to_remove} with VIF: {max_vif}")
        # Drop the feature with the highest VIF from the dataset
        X1 = X1.drop(columns=[feature_to_remove])
    
    # ELSE if no VIF is above the threshold, break the loop
    else:
        break

# Display result
final_vif_scores = calculate_vif(X1)
print(f"\n {final_vif_scores}")
```

### Step 2. Feature Significance Assessment (Backward Elimination)
To futher streamline our model and enhance its predictive power, we'll use backward elimination. This iterative process involves removing features that are statistical insignificant in predicting the length of stay. Significance will be assessed using p-values derived from our Logit regression models. Features with p-values exceeding 0.05 will be deemed non-significant and discarded (Srinidhi, 2019). By retaining only statistically significant features, we can ensure that the model includes only those variables that contribute meaningfully to explaining the dependent variable. Within this step, we'll also normalize the independent variables using MinMaxScaler. This will ensure that all features are on a comparable scale, preventing features with larger magnitudes from dominating the model (Brownlee, 2020).

```python
# Define backward elimination function with step summaries
def backward_elimination(independent_variables, dependent_variable, significance_level=0.05):
    
    # Set features list (all independent var names from df)
    features = independent_variables.columns.tolist()
    step = 1
    
    while len(features) > 0:
        # Add a constant to the features
        features_with_constant = sm.add_constant(independent_variables[features])
        
        # Fit the model and get p-values
        elimination_model = sm.Logit(dependent_variable, features_with_constant).fit() 
        p_values =  elimination_model.pvalues[1:]  # Excludes the intercept
        
        # Print a summary of the current step
        print(f"\n STEP {step}: LOGIT SUMMARY #{step}")
        print(elimination_model.summary())
    
        # Find the max p-value
        max_p_value = p_values.max()
        
        # Display the retained features and their corresponding p-values
        print(f"\nRETAINED VARIABLES + P-VALUES:\n{p_values}\n")
        
        # Check if the max p-value is greater than the significance level
        if max_p_value >= significance_level:
            excluded_feature = p_values.idxmax()
            features.remove(excluded_feature)
            print(f"REMOVED VARIABLE:\n{excluded_feature} (p-value: {max_p_value} exceeded 0.05 threshold)\n")
            print(f"--------------------------------------------------------------------------------------------------")
        else:
            print("No features were removed; all p-values are statistically significant at the 0.05 level.\n")
            break
        
        step += 1
    
    # Display the final selected features after backward elimination
    print(f"--------------------------------------------------------------------------------------------------")
    print(f"\nVARIABLES TO BE RETAINED IN THE FINAL MODEL:\n{features}")
    return features

# Normalize the independent variables (excluding the dependent variable)
independent_variables = df_subset.drop('diabetes', axis=1)  # Drop dependent variable column
normalized_df = pd.DataFrame(preprocessing.MinMaxScaler().fit_transform(independent_variables), columns=independent_variables.columns)

# Perform backward elimination on the normalized dataset
reduced_features = backward_elimination(normalized_df, df_subset['diabetes'])
```

Through this feature selection process, our goal is to create a more robust and interpretable regression model that accurately predicts length of stay based on a set of independent, statistically significant features. 

The features that will be retained in our reduced model include:

### Model Evaluation and Improvement

The initial logistic regression model included several variables, some of which were redundant or had minimal predictive power. To improve the model, we first removed gender_male and age due to multicollinearity concerns. After, we employed a backward elimination approach to identify and remove variables with statistically insignificant contributions. This iterative process involved removing the variable with the highest p-value, refitting the model, and repeating until all remaining variables had p-values below the chosen threshold of 0.05.

**To assess the relative improvement of the reduced model compared to the initial model, we'll use the Akaike Information Criterion (AIC) and the Likelihood Ratio Test (LR Test)**. These metrics provide a more comprehensive evaluation by considering both the model's fit to the data and its complexity.

Unlike the OLS summaries in Task 1, our Logit summaries do not directly provide AIC and LR Test values. Therefore, we'll need to calculate these metrics manually using NumPy and the metrics that our Logit summaries do provide:

![model_evaluation_metrics_task2.png](model_evaluation_metrics_task2.png)

```python
# Log-Likelihood values 
log_likelihood_initial = -5065.9  # Initial Model
log_likelihood_reduced = -5070.0  # Reduced Model

# Degrees of freedom 
df_initial = 15 
df_reduced = 7 

# No. Observations (dataset sample size)
n = 10000
```

### AIC Comparison

The Akaike Information Criterion (AIC) is a widely used metric for comparing the relative quality of statistical models. Lower AIC values suggest a better model fit, considering both the model's ability to explain the data and its complexity (Middleton, 2023). To calculate the AIC for both models, we can use the following general formula (Bevans, 2020):

$AIC = 2K - 2(\log L)$

Where:
- $K$ is the number of independent variables used in each model
<br>
- $\log L$ is the log-likelihood estimate provided by our Logit summaries (the likelihood that the model could have produced our observed y-values)

```python
# Calculate the AIC for both models
AIC_initial = 2 * df_initial - 2 * log_likelihood_initial
AIC_reduced = 2 * df_reduced - 2 * log_likelihood_reduced

print("AIC of the Initial Model:", AIC_initial)
print("AIC of the Reduced Model:", AIC_reduced)
```

The initial model exhibited an AIC of 10,161.8, while the reduced model achieved an AIC of 10,154.0. This represents a marginal improvement. The reduced model's slightly lower AIC suggests that it achieves a balance between model simplicity and fit that is ever-so-slightly better than the initial model.

### Likelihood Ratio Test (LRT)

To further compare the reduced model with the initial model, we’ll also use the likelihood ratio test (LRT). This statistical test directly assesses whether the reduced model is significantly worse or better than the initial model in terms of model fit. By examining the LRT statistic, we can determine if removing variables has made a significant impact on the model's performance.

A large LRT statistic suggests that the reduced model fits the data much worse than the initial model, indicating that the removed variables were important for explaining the variation in the data. Conversely, a small LRT statistic implies that the reduced model is similar in fit to the initial model, suggesting that the removed variables didn’t substantially contribute to the model's explanatory power.

To calculate the LRT statistic, we can use the following formula (White, 2017):

$LRT_{statistic} = -2 \left( \log L_{\text{reduced}} - \log L_{\text{initial}} \right)$

Where: 
- $\log L_{\text{reduced}}$ is the log-likelihood of the reduced model
- $\log L_{\text{initial}}$ is the log-likelihood of the initial model

This provides a statistical test to determine if the reduced model explains the dependent variable (diabetes) as well as the initial model. 

```python
# Calculate the Likelihood Ratio Test (LRT) Statistic 
LRT_statistic = -2 * (log_likelihood_reduced - log_likelihood_initial)

print("Likelihood Ratio Test Statistic:", LRT_statistic)
```

The LRT Statistic gives us a measure of how much the fit changed from our initial model to our reduced one. As we can see, our Likelihood Ratio Statistic was rather high at 8.2. However, this raw statistic alone doesn’t tell us whether the observed change is statistically significant. 

To assess the significance of the difference in fit, we can convert the LRT statistic into a p-value. This p-value represents the probability of observing an LRT statistic as extreme or more extreme (in favor of the reduced model) under the null hypothesis that the initial and reduced models have the same fit (Taboga, 2021).

Using the LRT Statistic, we can find the p-value of our Likelihood Ratio Test (Bevans, 2023): 

$LRT_{\text{p-value}} = 1 - e^{-\frac{LRT_{\text{statistic}}}{2}}$

Where: 
- ${LRT_{\text{statistic}}}$ is the LR test statistic we calculated above with the formula $LRT_{statistic} = -2 \left( \log L_{\text{reduced}} - \log L_{\text{initial}} \right)$
- $e$ is the base of the natural logarithm

This gives us a p-value approximation based on the relationship between the test statistic and the chi-square distribution.

```python
# Calculate the approximate p-value for the LR test
LRT_p_value = 1 - np.exp(-LRT_statistic / 2)

print("P-value of the LRT:", LRT_p_value)
```

The p-value of approximately 0.98 is quite high, which indicates that there is no statistically significant difference in fit between the initial and reduced models. This suggests that removing the variables in the reduced model did not meaningfully degrade the model's performance. 

### Summary
The Likelihood Ratio Test indicates no significant difference in the fit between the initial and reduced models, suggesting that the reduced model, despite its fewer variables, performs equally well. Moreover, the reduced model exhibits a slightly lower AIC, a metric that penalizes model complexity. Based on these results, the reduced model is preferable as it achieves comparable performance with a simpler structure. While the difference in performance between the two models is very small, the reduced model offers the advantage of greater interpretability and potentially better computational efficiency (since we’re working with less data).

#### General Formula
Our logistic regression model predicts the log-odds of the dependent variable (in this case, the likelihood of having diabetes) as a linear combination of the independent variables. Therefore, the general form of the logistic regression equation is (Datatab, 2024): <br>

$
\text{log-odds}(y) = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_n X_n
$

Where: <br>
- $\text{log-odds}(y)$ is the natural logarithm of the odds that the binary out come $y$ will occur 
- $\beta_0$ is the intercept (constant)
- $\beta_1 + \beta_2 + \dots + \beta_n$ are the regression coefficients of the independent variables 
- $X_1, X_2, \dots X_n$ are the independent variables

We derive our regression equation by using the coefficients provided in the Logit summary of our reduced model.

## D. Limitations of the Analysis
### Regression Modeling Limitations
**Sensitivity to Outliers** 

Logistic regression assumes that there are no extreme outliers or influential observations in the dataset. This is because logistic regression is sensitive to outliers and such data points can disproportionately impact the outcomes of the model, leading to biased and unreliable results. The presence of outliers or influential observations can make it difficult to interpret the results of our logistic regression model, causing misleading coefficient estimates, inflated standard errors, and unreliable p-values.

**Multicollinearity**

As we've previously seen, when independent variables are highly correlated, multicollinearity arises, inflating coefficient estimate variance and making it difficult to determine each predictor's individual effect. While not always problematic, high multicollinearity can destabilize model coefficients, making them "sensitive to small changes in the model." This can lead to "reduced precision of the estimated coefficients," weakening the statistical power of our regression model and resulting in large standard errors that may render our results meaningless (Frost, 2023).

**Assumption of Linearity in Log-Odds**

Logistic regression assumes that the relationship between the independent variables and the log-odds of the dependent variable is linear. This may not hold true for all the variables, especially those like daily_charge and length_of_stay, which might have non-linear relationships with diabetes. Non-linear relationships between the variables and the outcome might reduce the model’s accuracy and lead to misleading conclusions. 

**Correlation Does Not Imply Causation**

A significant relationship between our independent and dependent variables does not necessarily imply causation. Without controlled experimentation or additional context, our regression results can only show associations, not cause-and-effect relationships. Misinterpreting correlation as causation can result in incorrect conclusions.

### Data Preparation Limitations
**Inadequate Data Exploration**

Insufficient exploratory data analysis (EDA) can hinder our understanding of variable relationships, overlooking key features like non-linear relationships, outliers, or unusual patterns. If we've failed to thoroughly explore the data, this can result in poorly specified models that inaccurately capture variable relationships.

**Data Wrangling Errors**

Incorrect data transformations, such as encoding our categorical variables incorrectly or failing to normalize continuous variables, can lead to biased results or other errors.

### Model Reduction Method Limitations

**Overfitting** 

Using many predictors as we have done and having very large coefficients, like we do with variables like length_of_stay and daily_charge, can lead to overfitting, where the model captures noise rather than the underlying relationship, resulting in poor generalization to new data. Overfitted models may perform well on training data but fail on validation or test data, reducing their actual practical utility.

**Multicollinearity After Reduction**

Our model reduction techniques might not correctly resolve multicollinearity, especially if we don't account for variable correlations. Multicollinearity can still inflate standard errors and make it difficult for us to interpret the reduced model.
